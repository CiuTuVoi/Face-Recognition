```python
import numpy as np

# Đường dẫn đến file
names_path = "embeddings/names.npy"
embeds_path = "embeddings/embeddings.npy"

# Tải dữ liệu
names = np.load(names_path, allow_pickle=True)
embeddings = np.load(embeds_path)

# In danh sách hiện tại (tuỳ chọn)
print("Danh sách trước khi xóa:")
for i, name in enumerate(names):
    print(f"{i+1}: {name}")

# Tạo danh sách chỉ số cần xóa: từ index 2 đến 9 (ID 3 đến 10)
indices_to_delete = list(range(2, 57))

# Xoá phần tử tương ứng
names = np.delete(names, indices_to_delete, axis=0)
embeddings = np.delete(embeddings, indices_to_delete, axis=0)

# Lưu lại
np.save(names_path, names)
np.save(embeds_path, embeddings)

print("\n✅ Đã xóa các ID từ 3 đến 10 thành công.")
```
